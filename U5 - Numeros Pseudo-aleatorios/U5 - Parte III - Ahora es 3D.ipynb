### MatplotLib: Como Plotear en 3D

Hasta el momento, solo vimos como crear bellas gráficas de ploteos en 2 dimensiones, pero la vida real transcurre en 3D, y a veces nos gustaría visualizar estos fenómenos.<br>
Matplotlib nos permite crear gráficos en 3d con las mismas funciones que utilizabamos en 2d, si bien los argumentos son un poco distintos, si tienen este machete a mano no van a tener complicación alguna<br>

```python
import matplotlib.pyplot as plt
import numpy as np 

# por convensión, se suele llamar a los gráficos 3d como 'ax'

# para crear nuestro grafico y plotear sobre el distintos puntos en 3 dimensiones, debemos crear un 'subplot' con proyeccion en 3d y con los ejes en la misma escala
ax = plt.figure().add_subplot( projection='3d',aspect = 'equal'  )

# para luego, como haciamos en 2d,  plotear un scatterplot con los parametros 'x','y' y 'z': 
x = np.random.rand(200)* 2 - 1
y = np.random.rand(200)* 2 - 1
z = np.random.rand(200)* 2 - 1
ax.scatter(x,y,z, color = 'red') 
ax.set_xlim3d(-1, 1)
ax.set_ylim3d(-1, 1)
ax.set_zlim3d(-1, 1)

plt.show()
```

```python
ax = plt.figure().add_subplot( projection='3d',aspect = 'equal'  )

t = np.arange(0,100,0.1)
x = np.sin(t)
y = np.cos(t)
z = t
ax.set_xlim3d(-1, 1)
ax.set_ylim3d(-1, 1)
ax.set_zlim3d(-1, 100)

ax.plot(x,y,z, color = 'green') 
plt.show()
```

#### hora supongamos que queremos plotear vectores que simbolicen un campo vectorial.
Para ello, vamos a utiolizar la funcion 'quiver', que recibe como parámetros los puntos de la ubicacion, y el campo vectorial en si.



```python
#por ejemplo, supongamos un campo M = (-y/4,2x,-zy)
x = np.random.rand(300)* 2 - 1
y = np.random.rand(300)* 2 - 1
z = np.random.rand(300)* 2 - 1
u = -y/4
v = 2*x
w = z*(-y)
mag = np.linalg.norm([x,y,z])/75 #saco la magnitud para darle un largo a las flechitas, lo divido por 100 por que sino quedan enormes
ax = plt.figure().add_subplot( projection='3d',\
                              aspect = 'equal'  )
ax.set_xlim3d(-1, 1)
ax.set_ylim3d(-1, 1)
ax.set_zlim3d(-1, 1)

ax.quiver(x,y,z,u,v,w, length=mag, alpha = 0.5)
plt.show()
```

6. Calcule el volumen de 1/8 de esfera. de radio $r = 4m$ y $ z = \int_0^R \int_0^R \sqrt{R^{2}-x^{2}+y^{2}}dxdy $<br><br>
Para validar esta integral, tenga en cuenta que el volumen de 1/8 de esfera es <br><br>
$volumen =\frac{4}{3}\frac{\pi R^{3}}{8}$<br><br>

```python
### Tengamos en cuenta que no es necesario para obtener los resultados plotear cada punto, esto se hace para visualizar el método.
### Primero vamos a sacarlo numéricamente, luego vamos a plotearlo.

contador = 0
n = 4000
r = 4

for i in range (n):
    #como los nros son positivos, va a estar en el primer octante
    x = np.random.rand() * r
    y = np.random.rand() * r
    z = np.random.rand() * r
    d = x**2 + y**2 +z**2 

    if d <= r**2 :
            contador = contador + 1
            
integral = contador / n
vol_recinto = r**3
vol_mc = integral * vol_recinto
vol_def = round(((4/3) * np.pi * r**3 )/ 8,2) 
print('Vol montecarlo: '+ str(vol_mc) + ' m^3. Vol definicion: ' + str(vol_def) + 'm^3' )
```
